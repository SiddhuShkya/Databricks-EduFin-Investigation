# EduFin National Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
sns.set_theme(style="whitegrid")
```

```python
# Load the datasets
base_path = "../datasets/edufin_national/"
loans = pd.read_csv(base_path + "loans.csv")
customers = pd.read_csv(base_path + "customers.csv")
institutions = pd.read_csv(base_path + "institutions.csv")
dim_state = pd.read_csv(base_path + "dim_state.csv")
dim_city = pd.read_csv(base_path + "dim_city.csv")

print(f"Loaded {len(loans)} loans, {len(customers)} customers, {len(institutions)} institutions.")
```

```python
loans.head()
```

### Visualizations

```python
# Visualization 1: Loan Status Distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=loans, x='loan_status', palette='viridis')
plt.title('Distribution of Loan Statuses')
plt.xlabel('Loan Status')
plt.ylabel('Count')
plt.show()
```

```python
# Visualization 2: Customer Income Brackets
plt.figure(figsize=(8, 5))
sns.countplot(data=customers, x='income_bracket', palette='magma')
plt.title('Customer Income Brackets')
plt.xlabel('Income Bracket')
plt.ylabel('Count')
plt.show()
```

```python
# Visualization 3: Distribution of Loan Amounts
plt.figure(figsize=(10, 6))
sns.histplot(data=loans, x='loan_amount', bins=30, kde=True, color='blue')
plt.title('Distribution of Loan Amounts')
plt.xlabel('Loan Amount')
plt.ylabel('Frequency')
plt.show()
```
