```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
loans = spark.read.csv("../datasets/edufin_national/loans.csv", header=True)
customers = spark.read.csv("../datasets/edufin_national/customers.csv", header=True)
institutions = spark.read.csv("../datasets/edufin_national/loans.csv", header=True)
dim_state = spark.read.csv("../datasets/edufin_national/dim_state.csv", header=True)
dim_city = spark.read.csv("../datasets/edufin_national/dim_city.csv", header=True)
```

```python
loans.createOrReplaceTempView("loans")
customers.createOrReplaceTempView("customers")
institutions.createOrReplaceTempView("institutions")
dim_state.createOrReplaceTempView("dim_state")
dim_city.createOrReplaceTempView("dim_city")
```

Loans Information

---

```python
query = """
SELECT * FROM loans;
"""
spark.sql(query).show(10)
```

```python
query = """
SELECT COUNT(*) FROM loans;
"""
spark.sql(query).show(10)
```
